Código de atividade:

```python
"""
Simulador de Gerenciamento de Memória
Algoritmos de substituição de páginas: FIFO e LRU
"""


def simular_fifo(paginas, quantidade_frames, verbose=True):
    """FIFO: remove a página que está há mais tempo na memória.
    A lista 'memoria' mantém a ordem de chegada (posição 0 = mais antiga)."""
    memoria = []
    faltas = 0
    hits = 0

    for pagina in paginas:
        if pagina in memoria:
            hits += 1
            resultado = "HIT"
            # FIFO NÃO altera a ordem num HIT
        else:
            faltas += 1
            resultado = "PAGE FAULT"
            if len(memoria) >= quantidade_frames:
                memoria.pop(0)          # remove a mais antiga
            memoria.append(pagina)

        if verbose:
            print(f"Página {pagina:<3} -> {str(memoria):<25} {resultado}")

    return faltas, hits


def simular_lru(paginas, quantidade_frames, verbose=True):
    """LRU: remove a página menos recentemente usada.
    A lista 'memoria' mantém a ordem de uso (posição 0 = menos recente,
    última posição = mais recente)."""
    memoria = []
    faltas = 0
    hits = 0

    for pagina in paginas:
        if pagina in memoria:
            hits += 1
            resultado = "HIT"
            memoria.remove(pagina)      # LRU atualiza a recência num HIT:
            memoria.append(pagina)      # a página vai para o fim (mais recente)
        else:
            faltas += 1
            resultado = "PAGE FAULT"
            if len(memoria) >= quantidade_frames:
                memoria.pop(0)          # remove a menos recentemente usada
            memoria.append(pagina)

        if verbose:
            print(f"Página {pagina:<3} -> {str(memoria):<25} {resultado}")

    return faltas, hits


def mostrar_resumo(nome, faltas, hits):
    total = faltas + hits
    taxa_faltas = faltas / total * 100
    taxa_hits = hits / total * 100
    print(f"\n{nome}")
    print(f"Page Faults: {faltas}")
    print(f"Hits: {hits}")
    print(f"Taxa de Page Faults: {taxa_faltas:.2f}%")
    print(f"Taxa de Hits: {taxa_hits:.2f}%")
    return taxa_faltas, taxa_hits


def main():
    entrada = input("Digite as páginas separadas por espaço: ")
    paginas_usuario = [int(valor) for valor in entrada.split()]

    quantidade_frames = int(input("Digite a quantidade de frames: "))

    if not paginas_usuario or quantidade_frames < 1:
        print("Informe ao menos uma página e ao menos 1 frame.")
        return

    print("\nSequência:", paginas_usuario)
    print("Frames:", quantidade_frames)

    # --- FIFO ---
    print("\n" + "=" * 45)
    print("EXECUÇÃO FIFO")
    print("=" * 45)
    faltas_fifo, hits_fifo = simular_fifo(paginas_usuario, quantidade_frames)

    # --- LRU ---
    print("\n" + "=" * 45)
    print("EXECUÇÃO LRU")
    print("=" * 45)
    faltas_lru, hits_lru = simular_lru(paginas_usuario, quantidade_frames)

    # --- Resultados finais ---
    print("\n" + "=" * 45)
    print("RESULTADOS FINAIS")
    print("=" * 45)
    mostrar_resumo("FIFO", faltas_fifo, hits_fifo)
    mostrar_resumo("LRU", faltas_lru, hits_lru)

    # --- Comparação ---
    print("\n" + "=" * 45)
    print("COMPARAÇÃO")
    print("=" * 45)
    print(f"{'Algoritmo':<10}{'Page Faults':>12}{'Hits':>8}")
    print(f"{'FIFO':<10}{faltas_fifo:>12}{hits_fifo:>8}")
    print(f"{'LRU':<10}{faltas_lru:>12}{hits_lru:>8}")

    if faltas_fifo < faltas_lru:
        print(f"\nFIFO teve {faltas_lru - faltas_fifo} page fault(s) a menos que o LRU nesta sequência.")
    elif faltas_lru < faltas_fifo:
        print(f"\nLRU teve {faltas_fifo - faltas_lru} page fault(s) a menos que o FIFO nesta sequência.")
    else:
        print("\nOs dois algoritmos tiveram o mesmo número de page faults nesta sequência.")


if __name__ == "__main__":
    main()
```

## Respostas das questões:

**1. Qual é a diferença entre um HIT e um PAGE FAULT?**

Um HIT acontece quando a página solicitada já está carregada em algum frame da memória RAM, então o acesso é imediato. Um PAGE FAULT acontece quando a página não está na memória: o sistema operacional precisa buscá-la no armazenamento secundário e carregá-la em um frame livre ou, se a memória estiver cheia, substituir uma página já carregada.

**2. Qual é a principal diferença entre FIFO e LRU?**

O FIFO remove a página que **entrou primeiro** na memória (ordem de chegada), sem considerar se ela está sendo usada. O LRU remove a página que **está há mais tempo sem ser utilizada** (ordem de uso). Por isso, no FIFO um HIT não altera a ordem da fila, enquanto no LRU um HIT move a página para a posição de mais recente.

**3. Um page fault significa necessariamente que existe um erro no programa? Explique.**
Não. O page fault é um evento normal do funcionamento da memória virtual: só indica que a página ainda não está na RAM e precisa ser carregada. Ele sempre ocorre, por exemplo, nos primeiros acessos, quando a memória ainda está vazia. O problema só aparece quando os page faults são excessivos, porque deixam o sistema mais lento (no extremo, o *thrashing*).

**4. Aumentar a quantidade de frames sempre reduz o número de page faults no FIFO? Utilize o experimento da Anomalia de Belady para justificar.**
Não. Com a sequência 1, 2, 3, 4, 1, 2, 5, 1, 2, 3, 4, 5, o FIFO teve **9 page faults com 3 frames** e **10 page faults com 4 frames**. Ou seja, adicionar um frame piorou o desempenho. Isso é a Anomalia de Belady: como o FIFO decide apenas pela ordem de chegada, o conjunto de páginas na memória com 4 frames não contém necessariamente as páginas que estariam com 3 frames, e ele pode acabar removendo justamente as páginas que serão usadas logo em seguida.

**5. Em quais situações FIFO e LRU podem produzir resultados diferentes?**

Quando ocorre um HIT em uma página antiga que continua sendo usada. O FIFO ignora esse uso e a remove por ter entrado primeiro; o LRU a considera recente e remove outra. Exemplo da aula (sequência 1, 2, 3, 1, 4, 2, 5, 1 com 3 frames): após o HIT na página 1, o FIFO removeu a página 1 ao chegar a página 4, enquanto o LRU removeu a página 2. Com 3 frames o FIFO teve 6 faults e o LRU 7. Na sequência 7, 0, 1, 2, 0, 3, 0, 4, 2, 3, 0, 3, 2 aconteceu o contrário: FIFO 10 e LRU 9. Se a sequência não tiver nenhum HIT antes de a memória encher, ou se as páginas forem acessadas em ciclo maior que o número de frames, os dois tendem a dar o mesmo resultado.

**6. Por que o algoritmo Ótimo normalmente apresenta um bom resultado?**

Porque ele substitui a página que só será usada novamente **mais longe no futuro** (ou que nunca mais será usada). Assim, mantém na memória exatamente as páginas que serão necessárias primeiro, o que gera o menor número possível de page faults para aquela sequência. Na sequência clássica com 3 frames: FIFO 10, LRU 9 e Ótimo 7 page faults.

**7. Por que um sistema operacional real não consegue utilizar perfeitamente o algoritmo Ótimo?**

Porque ele exige conhecer antecipadamente toda a sequência futura de acessos às páginas, e o sistema operacional não sabe quais páginas os programas vão pedir depois (isso depende da execução, das entradas do usuário, de desvios no código etc.). Por isso o Ótimo serve como referência teórica para comparar outros algoritmos, e os sistemas reais usam aproximações como o LRU, que tenta prever o futuro com base no passado recente.
